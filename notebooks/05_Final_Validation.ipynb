```python
import pandas as pd

df = pd.read_csv("../data/master_dataset_FINAL_SUBMISSION_LOCKED.csv")

df.head()
```

```python
df['Price'].describe()
```

```python
metro_cities = [
    'Mumbai', 'Delhi', 'Bangalore', 'Chennai',
    'Hyderabad', 'Pune', 'Kolkata'
]

def realistic_price_flag(row):
    if row['City_clean'] in metro_cities and row['Price'] < 2000000:
        return False
    return True

df['price_realistic_flag'] = df.apply(realistic_price_flag, axis=1)
```

```python
df.groupby('City_clean')['price_realistic_flag'].value_counts()
```

```python
df = df[df['price_realistic_flag'] == True]
```

```python
df_app = df[df['price_realistic_flag'] == True].copy()
```

```python
df_app['price_lakh'] = (df_app['Price'] / 100000).round(2)
df_app['price_crore'] = (df_app['Price'] / 10000000).round(2)
```

```python
df_app.groupby('City_clean')['price_lakh'].describe()
```

```python
df_app.groupby('City_clean')['price_lakh'].describe()
```

```python
df_app.to_csv(
    "master_dataset_FINAL_APP_READY.csv",
    index=False
)
```
