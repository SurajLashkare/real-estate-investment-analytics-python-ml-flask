### Rental Yield Distribution Handling

In real estate markets, rental yields above **15–20%** are extremely rare and typically represent atypical or anomalous observations. While such values may exist in raw datasets, they are not representative of standard market behavior.

During exploratory data analysis, the rental yield distribution exhibited significant right-skewness driven by a small number of extreme outliers. To improve interpretability and ensure meaningful visualization, rental yield values were **capped at the 99th percentile** for EDA purposes.

This approach preserves overall dataset integrity while reducing the influence of extreme values, enabling clearer analysis of typical rental yield behavior without removing valid observations from the dataset.

```python
import numpy as np

top_cities = (
    df["City"]
    .value_counts()
    .head(5)
    .index
)

df["log_price"] = np.log1p(df["Price"])

plt.figure(figsize=(10,5))
sns.violinplot(
    data=df[df["City"].isin(top_cities)],
    x="City",
    y="log_price",
    inner="quartile"
)

plt.title("Log-Scaled Price Distribution Across Major Cities")
plt.ylabel("Log(Property Price)")
plt.show()
```

This transformation preserves all observations while improving interpretability and highlighting relative differences in market behavior across metropolitan regions.

```python
from scipy.stats import ttest_ind

high_price = df[df["Price"] > df["Price"].median()]["rental_yield_pct"]
low_price = df[df["Price"] <= df["Price"].median()]["rental_yield_pct"]

ttest_ind(high_price, low_price)
```

### Statistical Test: Rental Yield vs Property Price

An independent two-sample t-test was conducted to compare rental yields between high-priced and low-priced properties. The results indicate no statistically significant difference in mean rental yield between the two groups (p-value = 0.253).

This suggests that higher property prices do not necessarily translate into superior rental returns, highlighting the importance of yield-focused investment strategies rather than price-based assumptions.
